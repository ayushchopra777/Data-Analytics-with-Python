```python
# Create a list of different types of data
samples = [
    {
        "type": "structured",
        "value": {"Name": "Raj", "Age": 20}   # Dictionary
    },
    {
        "type": "semi-structured",
        "value": {"tags": ["python", "data"]}  # Dictionary with a list
    },
    {
        "type": "unstructured",
        "value": "A long free-text customer comment..."  # String
    }
]

# Loop through each item in the list
for s in samples:

    # Print the type of data and the Python data type
    print(f"{s['type']}: {type(s['value']).__name__}")
```
